### Dyanmic Foul Ball Heatmap Pipeline

This Jupyter Notebook walks though how to run the project, including the creation of heatmaps for multiple game scenarios. 

##### Importing Functions for Project

```python
from src.dataCollection import findFiles, hitChartData
from src.heatmapGeneration import filterData, drawHeatmap
```

##### Finalized Pipeline Function

```python
def pipeline(condition) -> None:
    """
    Pipeline for Foul Ball Heatmap

    Parameters
    ----------
    condition : str  
        Condition that specifies the game situation, including inning, runners on base, and batter handedness.

        Format: "Inning/RunnerOn1st/RunnerOn2nd/RunnerOn3rd/BatterHandedness"

        Fields:
        - Inning: Specifies The Current Inning and Half Inning
            - "T1" = Top of The 1st Inning
            - "B2" = Bottom of The 2nd Inning

        - RunnerOn1st / RunnerOn2nd / RunnerOn3rd: Indicates Whether Runner is Present on Each Base
        - "0 = No Runner on Base
        - "1" = Runner on Base

        - BatterHandedness: Specifies the Batters Handedness
        - "L" = Left Handed Batter
        - "R" = Right Handed Batter
            
    Returns
    -------
    None 
        Displays Matplotlib Visualization
    """
    
    # Finding File Names of All Teams
    parents = ["ANI", "ARN", "PHD", "VAS"]
    subfolders = ["ADQ", "EMB", "FET", "FXP", "IGL", "JPB", "MEX",  "MSE", "NLG", "OER", "QBR", "QHX", "SZJ", "WIF"]
    folderList = [f"{parent}/{sub}" for parent in parents for sub in subfolders]

    # Collecting Hit Chart Data
    data = hitChartData(folderList)

    # Filtering Data Based on Condition
    filteredData = filterData(data, condition)

    # Drawing Heatmap
    drawHeatmap(filteredData)
```

##### Example Game Scenario

**Pipeline Format:** `Inning/RunnerOn1st/RunnerOn2nd/RunnerOn3rd/BatterHandedness`

**Fields**:

Inning: Specifies The Current Inning and Half Inning
- `T1` = Top of The 1st Inning
- `B2` = Bottom of The 2nd Inning

RunnerOn1st / RunnerOn2nd / RunnerOn3rd: Indicates Whether Runner is Present on Each Base
- `0` = No Runner on Base
- `1` = Runner on Base

BatterHandedness: Specifies the Batters Handedness
- `L` = Left Handed Batter
- `R` = Right Handed Batter

| Pipeline Case | Inning | Runner on 1st | Runner on 2nd | Runner on 3rd | Batter Hand | Situation |
|---|---|---|---|---|---|---|
| `pipeline("T1/0/0/0/R")` | Top 1st | No | No | No | Right | Bases Empty |
| `pipeline("T1/1/0/0/L")` | Top 1st | Yes | No | No | Left | Runner on First |
| `pipeline("T1/0/0/0/L")` | Top 1st | No | No | No | Left | Bases Empty |
| `pipeline("T1/0/1/0/R")` | Top 1st | No | Yes | No | Right | Runner on Second |
| `pipeline("T1/1/0/1/L")` | Top 1st | Yes | No | Yes | Left | Runners on First and Third |
| `pipeline("T1/1/1/1/R")` | Top 1st | Yes | Yes | Yes | Right | Bases Loaded |

```python
# Bases Empty
pipeline("T1/0/0/0/R") 

# Runner on First After Single
pipeline("T1/1/0/0/L") 

# Bases Empty After Double Play
pipeline("T1/0/0/0/L")  

# Runner on Second After Double
pipeline("T1/0/1/0/R")

# Runners on First and Third After Single
pipeline("T1/1/0/1/L") 

# Bases Loaded
pipeline("T1/1/1/1/R")
```
